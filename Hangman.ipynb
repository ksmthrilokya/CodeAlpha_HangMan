```python
import random

def play_hangman():
    # Step 1: Predefined list of 5 words
    words = ["python", "hangman", "internship", "coding", "developer"]

    # Step 2: Pick a random word
    word = random.choice(words)

    # Step 3: Track guessed letters and wrong attempts
    guessed_letters = []
    wrong_guesses = 0
    max_wrong_guesses = 6

    print("Welcome to Hangman!")
    print(f"The word has {len(word)} letters.")

    # Step 4: Main game loop
    while wrong_guesses < max_wrong_guesses:
        # Display current progress: correct letters shown, others as _
        display_word = ""
        for letter in word:
            if letter in guessed_letters:
                display_word += letter + " "
            else:
                display_word += "_ "
        print("\nWord: " + display_word)
        print(f"Wrong guesses: {wrong_guesses}/{max_wrong_guesses}")

        # Check for win
        if "_" not in display_word:
            print("\nCongratulations! You guessed the word: " + word)
            break

        # Step 5: Get player's guess
        guess = input("Guess a letter: ").lower()

        # Basic input validation
        if len(guess) != 1 or not guess.isalpha():
            print("Please enter a single letter.")
            continue

        if guess in guessed_letters:
            print("You already guessed that letter.")
            continue

        guessed_letters.append(guess)

        # Step 6: Check if guess is correct
        if guess in word:
            print(f"Good guess! '{guess}' is in the word.")
        else:
            wrong_guesses += 1
            print(f"Wrong guess! '{guess}' is not in the word.")

    # Step 7: Check if player lost
    if wrong_guesses == max_wrong_guesses:
        print(f"\nGame Over! You've used all {max_wrong_guesses} wrong guesses.")
        print(f"The word was: {word}")


if __name__ == "__main__":
    play_hangman()
```
